## Q1. Mental Health and Income Group

The 2011 Behavioral Risk Factor Surveillance System (BRFSS) contains individual-level data with the following variables:

- `MENTHLTH` -- days of Poor Mental Health in the Past 30 Days
- `INCOME2` -- income category (1 = less than \$10,000, ..., 8 = \$75,000 or more)

Create a binary variable `under_10k` equal to 1 if `INCOME2 == 1` and 0 otherwise. Run a linear regression such that the mental health variable is the outcome variable and `under_10k` is the independent variable.

````python
# Load data
data = pd.read_csv(file_path + 'BRFSS_cleaned.csv')

# Create binary variable
data['under_10k'] = (data['INCOME2'] == 1).astype(int)
data.head()
````

````python
# Regression: MENTHLTH ~ under_10k
model = smf.ols('MENTHLTH ~ under_10k', data=data).fit()
print(model.summary().tables[1])
````

a) What does $\hat\beta_0$ represent in context?

$\hat\beta_0$ is the average number of poor mental health days for individuals who are *not* in the under-\$10,000 income group (`under_10k = 0`). It represents the baseline level of poor mental health for the reference group -- everyone earning \$10,000 or more.

b) What does $\hat\beta_1$ represent in context?

$\hat\beta_1$ is the difference in the average number of poor mental health days between individuals earning under \$10,000 and those earning \$10,000 or more. A positive coefficient means that individuals in the lowest income group report more days of poor mental health on average than those with higher incomes.

## Q2. Reading Regression Output

A researcher wants to test whether having a college degree is associated with higher hourly wages. They code a binary variable `college` equal to 1 if the worker has a college degree and 0 otherwise, then estimate the following model:

$$\texttt{wage}_i = \beta_0 + \beta_1 \cdot \texttt{college}_i + \varepsilon_i$$

Their regression output shows:

```
                 coef    std err          t      P>|t|      [0.025      0.975]
------------------------------------------------------------------------------
Intercept       18.42      0.310     59.419      0.000      17.81      19.03
college          9.65      0.440     21.932      0.000       8.79      10.51
------------------------------------------------------------------------------
```

a) Interpret the intercept coefficient (18.42) in context.

The intercept (18.42) is the predicted average hourly wage for workers without a college degree (`college = 0`). Workers without a college degree earn an estimated \$18.42 per hour on average.

b) Interpret the slope coefficient (9.65) in context.

The coefficient on `college` (9.65) tells us that workers with a college degree earn an estimated \$9.65 more per hour than workers without a college degree, on average. The predicted wage for college-educated workers is $18.42 + 9.65 = \$28.07$ per hour.

c) What is the null hypothesis being tested for the `college` coefficient?

$H_0: \beta_1 = 0$. There is no difference in average hourly wages between workers with and without a college degree.

d) Interpret the p-value in words.

The p-value is approximately 0.000, meaning that if there were truly no wage difference between college-educated and non-college-educated workers ($\beta_1 = 0$), the probability of observing a difference as large as \$9.65 (or larger) by chance alone is essentially zero. We reject the null hypothesis and conclude that having a college degree is associated with a statistically significant difference in hourly wages.
