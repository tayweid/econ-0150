## Q1. Unemployment and BMI

The article [*Zhang, Q., Lamichhane, R., & Wang, Y. (2014)*](https://pmc.ncbi.nlm.nih.gov/articles/PMC4449673/) in the Journal of Clinical Medicine, 3(1), 153-166 studies the associations between US adult obesity and mental health and state/county economic conditions in recessions.

a) (*optional: no need to answer this question*) Reading through the introduction, what is the primary research question the authors are trying to answer?

The authors investigate whether local economic conditions during the Great Recession -- specifically county-level unemployment rates -- are associated with changes in adult obesity and mental health outcomes.

b) (*optional: no need to answer this question*) Reading through the methods section, what dataset(s) do the authors use for their analysis? Be specific about the years and key variables.

The authors use the Behavioral Risk Factor Surveillance System (BRFSS) from 2005-2011, merged with county-level unemployment data from the Bureau of Labor Statistics. Key variables include BMI, self-reported mental health days, and county unemployment rates.

c) (*optional: no need to answer this question*) What are the main outcome and predictor variables?

The main outcome variables are BMI and days of poor mental health. The main predictor variable is county-level unemployment rate.

d) You are provided with a sample of individual-level data from the 2011 Behavioral Risk Factor Surveillance System (BRFSS), which includes the following variables:

- `BMI5` -- individual Body Mass Index (kg/m²)
- `MENTHLTH` -- days of Poor Mental Health in the Past 30 Days
- `unemployment_rate` -- county-level unemployment rate (in percentage points)

Using this data, estimate the following linear model:

$$\texttt{BMI5}_i = \beta_0 + \beta_1 \cdot \texttt{unemployment\_rate}_i + \varepsilon_i$$

Report the estimated coefficients $\hat\beta_0$ and $\hat\beta_1$, and $p$-values.

````python
# Load Data
data = pd.read_csv(file_path + 'BRFSS_cleaned.csv')
data.head()
````

````python
# Regression: BMI5 ~ unemployment_rate
model_q1 = smf.ols('BMI5 ~ unemployment_rate', data=data).fit()
print(model_q1.summary().tables[1])
````

e) State the null hypothesis for $\beta_1$.

The null hypothesis is $H_0: \beta_1 = 0$. That is, there is no linear relationship between county-level unemployment rate and individual BMI.

f) Interpret your estimated coefficient $\hat\beta_1$ in words. What does the estimate suggest about the relationship between county unemployment and individual BMI?

The estimated coefficient $\hat\beta_1$ tells us that a one percentage point increase in county-level unemployment rate is associated with a small increase in individual BMI. The p-value is very small, so the relationship is statistically significant. This suggests that individuals living in counties with higher unemployment tend to have slightly higher BMI, consistent with the idea that economic hardship may be associated with worse health outcomes.

## Q2. Unemployment and Mental Health

The 2011 BRFSS also includes information on mental health. The variable `MENTHLTH` records the number of days in the past 30 days when the respondent's mental health was "not good." Estimate the following linear model:

$$\texttt{MENTHLTH}_i = \beta_0 + \beta_1 \cdot \texttt{unemployment\_rate}_i + \varepsilon_i$$

a) Report and interpret the estimated coefficient $\hat\beta_1$.

````python
# Regression: MENTHLTH ~ unemployment_rate
model_q2 = smf.ols('MENTHLTH ~ unemployment_rate', data=data).fit()
print(model_q2.summary().tables[1])
````

The estimated coefficient $\hat\beta_1$ tells us that a one percentage point increase in county-level unemployment rate is associated with an increase in the number of days of poor mental health reported in the past 30 days. The relationship is statistically significant. This suggests that individuals in counties with higher unemployment tend to report worse mental health.

b) Write a complete interpretation of the p-value for $\hat\beta_1$ from your model in part (a). What exactly does it tell us in words?

The p-value tells us the probability of observing a coefficient as large or larger than $\hat\beta_1$ (in absolute value) if the null hypothesis were true -- that is, if there were truly no relationship between county unemployment and mental health days ($\beta_1 = 0$). A very small p-value means that it would be extremely unlikely to see a relationship this strong by chance alone, so we reject the null hypothesis and conclude that there is a statistically significant relationship between county unemployment and self-reported mental health.

The intercept serves as a mathematical anchor for the regression line but should not be interpreted literally.

For each additional degree Fahrenheit in temperature, the model predicts an increase of $4.78 in daily ice cream sales. Warmer days are associated with higher sales.

d) What is the null hypothesis being tested for the `temperature` coefficient?

$H_0: \beta_1 = 0$. There is no linear relationship between temperature and daily ice cream sales.

e) Interpret the p-value for the `temperature` coefficient in words.

The p-value is approximately 0.000, meaning that if there were truly no relationship between temperature and ice cream sales ($\beta_1 = 0$), the probability of observing a slope as large as 4.78 (or larger) by chance alone is essentially zero. We reject the null hypothesis and conclude that temperature is a statistically significant predictor of ice cream sales.
