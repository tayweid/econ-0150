# Exercise 1.1 | Categorical Variables

Lets visualize some categorical variables.

```python
# Imports
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# File Path
file_path = 'https://econ-0150.tayweid.io/data/'
```

## Dataset 1: Coffee_Shops.csv

```python
shops = pd.read_csv(file_path + 'Coffee_Shops.csv')
```

```python
# Summary Table
shops.value_counts()
```

```python
# Countplot (bar plot)
sns.countplot(data=shops, y='STATE', hue='STATE')
```

```python
# Create a binary categorical variable
shops['CA'] = np.where(shops['STATE'] == 'CA', 'CA', 'Other')
```

```python
# Countplot
sns.countplot(data=shops, y='CA', hue='CA')
```

## Dataset 2: employment_status.csv

```python
status = pd.read_csv(file_path + 'employment_status.csv')
```

```python
status['Employment Status'].value_counts()
```

```python
sns.countplot(data=status, y='Employment Status', hue='Employment Status')
```

## Dataset 3: employment_sector.csv

```python
sector = pd.read_csv(file_path + 'employment_sector.csv')
```

```python
sector['Sector'].value_counts()
```

```python
sns.countplot(data=sector, y='Sector', hue='Sector')
```

## Dataset 4: economic_optimism.csv

```python
optimism = pd.read_csv(file_path + 'economic_optimism.csv')
```

```python
# Order The Categories
order = [
    'Very Pessimistic',
    'Somewhat Pessimistic',
    'Neutral',
    'Somewhat Optimistic',
    'Very Optimistic'
]
```

```python
optimism['Economic Optimism'].value_counts().reindex(order)
```

```python
sns.countplot(data=optimism, y='Economic Optimism', hue='Economic Optimism', order=order)
```
