````python
# Imports
import pandas as pd
import statsmodels.formula.api as smf
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# File Path
file_path = 'https://tayweid.github.io/econ-0150/data/'
data = pd.read_csv(file_path + 'BRFSS_cleaned.csv')
data.head()
````

#### Q1. Revisiting the Recession and Health

**a)** Write a regression model.

$$\texttt{BMI5}_i = \beta_0 + \beta_1 \times \texttt{unemployment\_rate}_i + \beta_2 \times \texttt{Female}_i + \varepsilon_i$$

**b)** Fit the simple fixed-effect model.

````python
model_b = smf.ols('BMI5 ~ Female', data=data).fit()
print(model_b.summary().tables[1])
````

$\hat\beta_0$ is the average BMI for males (the reference group, when Female = 0). $\hat\beta_1$ is the difference in average BMI between females and males. If negative, females have lower average BMI than males.

**c)** Fit the model from part (a).

````python
model_c = smf.ols('BMI5 ~ unemployment_rate + Female', data=data).fit()
print(model_c.summary().tables[1])
````

- $\hat\beta_0$: predicted BMI for a male in a county with 0% unemployment
- $\hat\beta_1$: a 1 percentage point increase in county unemployment is associated with a $\hat\beta_1$ unit change in BMI, holding gender constant
- $\hat\beta_2$: females have $\hat\beta_2$ units different BMI than males on average, holding unemployment constant

**d)** The coefficient on `unemployment_rate` should be very similar to the simple regression in HW 4.1. This tells us that gender is not a major confounding variable — it does not substantially change the estimated relationship between unemployment and BMI.

**e)** The model assumes that the effect of unemployment on BMI is the same for men and women. The two lines (one for males, one for females) are parallel — they have the same slope ($\beta_1$) but different intercepts. This means unemployment shifts BMI by the same amount regardless of gender.
