#### Q1. Reading Regression Output with Interactions

```
                       coef    std err          t      P>|t|      [0.025      0.975]
------------------------------------------------------------------------------------
Intercept             88.50      3.200     27.656      0.000      82.20      94.80
class_size            -0.95      0.140     -6.786      0.000      -1.22      -0.68
private                5.40      5.100      1.059      0.290      -4.63      15.43
class_size:private     0.60      0.280      2.143      0.033       0.05       1.15
------------------------------------------------------------------------------------
```

**a)** The predicted test score for a public school with class size 0 is 88.50. This is the intercept ($\hat\beta_0$).

**b)** For public schools, one additional student is associated with a 0.95-point decrease in test scores. This is $\hat\beta_1 = -0.95$.

**c)** For private schools, the effect is $\hat\beta_1 + \hat\beta_3 = -0.95 + 0.60 = -0.35$. One additional student is associated with a 0.35-point decrease in test scores for private schools.

**d)** The coefficient on `private` (5.40) represents the predicted difference in test scores between private and public schools when class size is 0. It is not statistically significant (p = 0.290), so we cannot conclude that private schools score differently from public schools at a class size of zero.

**e)** The interaction coefficient (0.60) tells us that the negative effect of class size is 0.60 points smaller for private schools than for public schools. Private schools are less harmed by larger class sizes. This is statistically significant (p = 0.033), so there is evidence that the class size effect differs by school type.

#### Q2. Do Recessions Affect Men and Women Differently? (*optional*)

````python
# Imports
import pandas as pd
import statsmodels.formula.api as smf
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# File Path
file_path = 'https://econ-0150.tayweid.io/data/'
data = pd.read_csv(file_path + 'BRFSS_cleaned.csv')
data.head()
````

**a)** Write a regression model with both a fixed effect and an interaction term.

$$\texttt{BMI5}_i = \beta_0 + \beta_1 \times \texttt{unemployment\_rate}_i + \beta_2 \times \texttt{Female}_i + \beta_3 \times \texttt{unemployment\_rate}_i \times \texttt{Female}_i + \varepsilon_i$$

**b)** Fit the interaction model without the fixed effect.

````python
model_b = smf.ols('BMI5 ~ unemployment_rate + unemployment_rate:Female', data=data).fit()
print(model_b.summary().tables[1])
````

- For males: the effect of a 1 pp increase in unemployment on BMI is $\hat\beta_1$
- For females: the effect is $\hat\beta_1 + \hat\beta_2$

**c)** Fit the full model.

````python
model_c = smf.ols('BMI5 ~ unemployment_rate + Female + unemployment_rate:Female', data=data).fit()
print(model_c.summary().tables[1])
````

- $\hat\beta_0$: predicted BMI for a male in a county with 0% unemployment
- $\hat\beta_1$: for males, a 1 pp increase in unemployment is associated with a $\hat\beta_1$ change in BMI
- $\hat\beta_2$: at 0% unemployment, females have $\hat\beta_2$ units different BMI than males
- $\hat\beta_3$: the difference in the unemployment effect between females and males — for females, the effect of a 1 pp increase in unemployment is $\hat\beta_1 + \hat\beta_3$

**d)** Predictions:

````python
b0, b1, b2, b3 = model_c.params

print("Male, 5% unemp: ", round(b0 + b1*5 + b2*0 + b3*5*0, 2))
print("Female, 5% unemp:", round(b0 + b1*5 + b2*1 + b3*5*1, 2))
print("Male, 10% unemp: ", round(b0 + b1*10 + b2*0 + b3*10*0, 2))
print("Female, 10% unemp:", round(b0 + b1*10 + b2*1 + b3*10*1, 2))
````

**e)** If $\hat\beta_3$ is large and statistically significant, the effect of unemployment on BMI differs by gender — the parallel lines assumption from HW 5.1 doesn't hold. If $\hat\beta_3$ is small or insignificant, the simpler model with parallel lines may be sufficient.
