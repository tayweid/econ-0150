# Exercise 2.1 | Bivariate Relationships

```python
# Imports
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Load Data
file_path = 'https://econ-0150.tayweid.io/data/'
```

## Anscombe's Quartet

These four datasets have the same mean, standard deviation, and correlation.

```python
# Load Anscombe's quartet (built into seaborn)
anscombe = sns.load_dataset('anscombe')
anscombe.head()
```

```python
# Summary statistics by dataset
anscombe.groupby('dataset').agg({'x': ['mean', 'std'], 'y': ['mean', 'std']})
```

```python
# Visualize all four datasets
sns.relplot(anscombe, x='x', y='y', col='dataset')
```

## Coffee Production and GDP

Was the relationship between coffee production and GDP different in 1980?

```python
gdp = pd.read_csv(file_path + 'Beans_GDP_1980.csv', index_col=0)
gdp.head()
```

```python
# Log both x and y variables
gdp['log_GDP'] = np.log(gdp['GDP'])
gdp['log_prod'] = np.log(gdp['coffee_prod'])

# Plot the log variables
sns.scatterplot(gdp, x='log_GDP', y='log_prod')
```

{++No. In 1980 too,++} countries with larger GDP tend to produce more coffee.

## Coffee and Oil Prices

But are the two prices positively or negatively related to each other?

```python
prices = pd.read_csv(file_path + 'Coffee_Oil.csv')
prices.head()
```

```python
# Scatterplot
sns.scatterplot(prices, x='Coffee', y='Oil')
```
