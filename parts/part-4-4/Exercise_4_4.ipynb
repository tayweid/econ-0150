#### Model 3 | First Differences (X, Y)

```python
# Step 1. Create first differences of X
data['unemployment_diff'] = data['unemployment'].diff()
data = data.dropna()

# Step 2. Fit the double differences model
model3 = smf.ols('gdp_diff ~ unemployment_diff', data=data).fit()
print(model3.summary().tables[1])
```

```python
# Step 3. Residual plot and lagged residual plot
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

ax1.scatter(model3.predict(), model3.resid, alpha=0.3)
ax1.axhline(0, color='red', linestyle='--')
ax1.set_xlabel('Predicted')
ax1.set_ylabel('Residual')
ax1.set_title('Residual Plot')

ax2.scatter(model3.resid.values[:-1], model3.resid.values[1:], alpha=0.3)
ax2.axhline(0, color='black', linestyle='--', alpha=0.3)
ax2.axvline(0, color='black', linestyle='--', alpha=0.3)
ax2.set_xlabel('Residual(t-1)')
ax2.set_ylabel('Residual(t)')
ax2.set_title('Lag Plot')

sns.despine()
plt.tight_layout()
```

#### Model 4 | Growth Rates

```python
# Step 1. Calculate growth rates
data['gdp_growth'] = data['gdp'].pct_change()
data['unemployment_growth'] = data['unemployment'].pct_change()
data = data.dropna()

# Step 2. Fit the growth rate model
model4 = smf.ols('gdp_growth ~ unemployment_growth', data=data).fit()
print(model4.summary().tables[1])
```

```python
# Step 3. Residual plot and lagged residual plot
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

ax1.scatter(model4.predict(), model4.resid, alpha=0.3)
ax1.axhline(0, color='red', linestyle='--')
ax1.set_xlabel('Predicted')
ax1.set_ylabel('Residual')
ax1.set_title('Residual Plot')

ax2.scatter(model4.resid.values[:-1], model4.resid.values[1:], alpha=0.3)
ax2.axhline(0, color='black', linestyle='--', alpha=0.3)
ax2.axvline(0, color='black', linestyle='--', alpha=0.3)
ax2.set_xlabel('Residual(t-1)')
ax2.set_ylabel('Residual(t)')
ax2.set_title('Lag Plot')

sns.despine()
plt.tight_layout()
```
